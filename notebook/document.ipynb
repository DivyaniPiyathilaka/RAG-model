```python
## Create a simple txt file
import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
sample_text = {
    "../data/text_files/python_intro.txt": """This is the content of document one. It contains information about Python programming.", ##file 1
     
    python is a versatile programming language used for web development, data analysis, artificial intelligence, and more.""",

    "../data/text_files/machine_learning.txt": """This is the content of document two. It contains information about Data Science.""", ##file 2
}

for file_path, content in sample_text.items():
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(content)

print("Sample text files created.")
```

```python
### Read files using text loader
from langchain_community.document_loaders import TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt" , encoding="utf-8")
document = loader.load()
print(document)
```

```python
### Directory Loader
from langchain_community.document_loaders import DirectoryLoader

## Load(read) all .txt files from the directory
dir_loader = DirectoryLoader(
    "../data/text_files", 
    glob="**/*.txt",  ##pattern to match files
    loader_cls=TextLoader,   ##specify the loader class to use for loading files 
    loader_kwargs={'encoding': 'utf-8'},  ##additional arguments for the loader class
    show_progress=False  
)

documents = dir_loader.load()
documents
```

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

###reading pdf files 

dir_loader = DirectoryLoader(
    "../data/pdf", 
    glob="**/*.pdf",  ##pattern to match files
    loader_cls=PyMuPDFLoader,   ##specify the loader class to use for loading files 
    show_progress=False  
)

pdf_documents = dir_loader.load()
pdf_documents
```
